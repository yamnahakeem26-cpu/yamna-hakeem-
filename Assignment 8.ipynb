```python
class Employee():
    def __init__(self,name,number,address,salary,post,year_of_joining,education,year_of_experience):
        self.name=name
        self.number=number
        self.address=address
        self.salary=salary
        self.post=post
        self.year=year_of_experience
        self.ed=education
        self.y=year_of_joining

    def describe_employee(self):
        print(f'I am {self.name} , my number is {self.number},my address is {self.address},my salary is {self.salary} ,my year of joining is {self.y} , my Ededucation is {self.ed},my year of experience is {self.y}.')

    def update_year_of_experience(self,update_year):
        self.uy=update_year
        self.y+=self.uy
        print(f'{self.name} update year of experience is {self.y}')
    def increment(self,update_salary):
        self.u=update_salary
        self.salary+=self.u
        print(f'now {self.name} salary is {self.salary}')


#instance
employee101=Employee("Yamna",555,"A-430,saadi town",90000,'Manager',2017,"M.Sc. in Zoology",8)
print(employee101.address)      #print Attributes
print(employee101.y)
print(employee101.name)
print(employee101.ed)
#call methods
employee101.describe_employee()
employee101.update_year_of_experience(2)
employee101.increment(2000)


#10instance
E1=Employee('Janita',111,'A-456,liaquatabad karachi',120000,'CEO',2000,'Intermediate',26)
E2=Employee('AbdulRehman',222,"A-555,Malir Cant",444444,'Supervisor',2015,'B.Sc',11)
E3=Employee('Misbah',333,"A-666,Malir Halt",300000,'Owner',1999,'Intermediate',25)
E4=Employee('Sunila',444,'B-430,Saadi Garden',100000,'Manager',2024,'M.Sc',7)
E5=Employee('Abdul Aleem',555,'C=436,Kareemabad',50000,'Teacher',2024,'B.Sc',3)
E6=Employee('Rabbia',666,'Block 7,North Nazimabad',80000,'Admin',2023,'B.com',5)
E7=Employee('Adeeena',777,'A-456,liaquatabad karachi',55000,'Finance',2024,'Intermediate',5)
E8=Employee('Tehreem',888,'B-430,Saadi Garden',100000,'Manager',2024,'M.Sc',7)
E9=Employee("Nabira",999,"A-666,Malir Halt",300000,'Owner',1999,'Intermediate',25)
E10=Employee('Binish',000,'B-430,Saadi Garden',100000,'Manager',2024,'M.Sc',7)

##Employee's detail

E1.describe_employee()
E2.describe_employee()
E3.describe_employee()
E4.describe_employee()
E5.describe_employee()
E6.describe_employee()
E7.describe_employee()
E8.describe_employee()
E9.describe_employee()
E10.describe_employee()


#each employee update year of experience

E1.update_year_of_experience(2)
E2.update_year_of_experience(2)
E3.update_year_of_experience(5)
E4.update_year_of_experience(2)
E5.update_year_of_experience(4)
E6.update_year_of_experience(2)
E7.update_year_of_experience(2)
E8.update_year_of_experience(3)
E9.update_year_of_experience(2)
E10.update_year_of_experience(6)

#each employee's salary increment

E1.increment(3000)
E2.increment(2000)
E3.increment(1000)
E4.increment(2000)
E5.increment(5000)
E6.increment(2000)
E7.increment(1000)
E8.increment(6000)
E9.increment(2000)
E10.increment(1500)
    
        
```
